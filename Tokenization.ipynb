```python
# Tokenization Workbook: Word, Subword, and Character Levels

# This Jupyter workbook demonstrates different tokenization strategies
# using a sample paragraph. It also includes an exercise cell where
# students can experiment with different languages.

# ------------------------------------------------------------
# Cell 1: Setup
# ------------------------------------------------------------

# Install required libraries (run once)
!pip install nltk sentencepiece transformers

import nltk
from nltk.tokenize import word_tokenize

# Download resources (run once)
nltk.download('punkt')

from transformers import AutoTokenizer
import sentencepiece as spm
```

```python
# ------------------------------------------------------------
# Cell 2: Input Paragraph
# ------------------------------------------------------------

paragraph = """Natural Language Processing (NLP) enables computers to understand,
interpret, and generate human language. Tokenization is a crucial
preprocessing step in most NLP pipelines.
"""
print(paragraph)
```

```python
import nltk
nltk.download('punkt_tab')

# ------------------------------------------------------------
# Cell 3: Word-Level Tokenization
# ------------------------------------------------------------
word_tokens = word_tokenize(paragraph)
print("Word-level tokens:")
print(word_tokens)
```

```python
# ------------------------------------------------------------
# Cell 4: Character-Level Tokenization
# ------------------------------------------------------------
char_tokens = list(paragraph)
print("Character-level tokens:")
print(char_tokens)
```

```python
# ------------------------------------------------------------
# Cell 5: Subword Tokenization (BPE / WordPiece)
# ------------------------------------------------------------

# Using a pretrained tokenizer (WordPiece-based BERT tokenizer)
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
subword_tokens = tokenizer.tokenize(paragraph)

print("Subword-level tokens:")
print(subword_tokens)
```

```python
# ------------------------------------------------------------
# Cell 6: Comparison Summary
# ------------------------------------------------------------
print("Number of word tokens:", len(word_tokens))
print("Number of subword tokens:", len(subword_tokens))
print("Number of character tokens:", len(char_tokens))
```
